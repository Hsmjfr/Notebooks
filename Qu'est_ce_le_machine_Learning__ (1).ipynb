# L'analogie de la pizza 🍕

Imaginez avoir un ami qui n'a jamais vu de pizza de sa vie. Vous souhaitez lui apprendre à les reconnaître. Comment procédez vous ?

1) Lui apprendre avec des règles précises

   Une pizza est une pâte ronde avec de la sauce tomate et du fromage fondu.

   Mais qu'en est il des pizzas qui ne sont pas rondes, celle à la crème, celle aux anchois ? Il faudrait définir toutes ces règles à la main   l'une après l'autre. C'est un travail interminable.

2) Lui apprendre avec des exemples

   Vous lui donnez 1000 photos de pizzas et 1000 photos de choses qui ne sont pas des pizzas. Puis, vous le laissez apprendre à reconnaitre ce qu'est une pizza tout seul.


Le Machine Learning, c'est **exactement ça**.

On apprend à un ordinateur à apprendre, avec des exemples plutôt qu'avec des règles fixes et c'est extrêmement puissant.

C'est très puissant parce que :
- Les règles sont souvent **trop complexes** ou **inconnues** (ex : qu'est ce qu'un spam ?)
- Les règles **changent avec le temps** (le spam évolue)
- Un programme classique ne **s'améliore pas** avec l'expérience — un modèle de ML, peut le faire

| Programmation classique | Machine Learning |
|---|---|
| On écrit les règles | On donne des exemples |
| "Si X alors Y" | "Voici 1000 cas, apprends tout seul" |
| Rigide | S'adapte |

# Les 3 Grandes Familles

On distingue 3 grandes familles au sein du Machine Learning.

### 1) L'apprentissage supervisé :

> On construit un modèle qui a accès à des exemples avec les bonnes réponses. Par exemple on veut un algorithme qui sache différencier les pizzas des brownies. Il aura accès à un ensemble d'images de brownies et de pizzas accompagnées à chaque fois de la bonne réponses.

Le modèle apprend tout seul à les différencier en se servant des bonnes réponses fournies.

2 sous-types :

- Classification : prédire une catégorie (spam/non spam, pizza/gateau ...)
- Régression : prédire un nombre (prix d'un actif, la pluviométrie, la température ...)

Exemples concrets :
- Prédire si un email est un sapm ou non
- Prédire le prix d'une maison

### 2) L'apprentissage par renforcement :

> On donne au modèle des objectifs et on le **récompense** quand il les atteint. Il finit donc par apprendre le comportement qu'il doit adopter et s'améliore avec le temps.

Si ce n'est pas clair, pensez au dressage d'un animal. On veut qu'il adopte un bon comportement. Donc à chaque bonne action on lui donne une récompense et à chaque mauvaise action rien.

Exemples concrets:
- Une IA qui apprend à jouer aux échecs
- Une IA qui apprend à garer une voiture

### 3) L'apprentissage non supervisé :

> On donne des données au modèle et on le laisse tout seul trouver les **structures cachés**.

Exemples concrets:
- Regrouper des articles de presses par thème
- Identifier des segments de client notamment dans le cas de projet Marketing

## 🏠 Analogie du prix d'une maison

Imaginez que vous êtes agent immobilier. Vous voulez prédire le prix d'une maison en fonction de sa surface, son nombre de chambres, son quartier...

### 📊 Les "features" (variables, caractéristiques)

> Ce sont les **informations** qu'on donne au modèle.

Pour une maison : surface, nombre de pièces, distance du centre-ville, année de construction...

**Analogie :** c'est la fiche descriptive de la maison. Plus elle est complète, mieux on peut prédire.

### 🎯 La "target" (cible, étiquette)

> C'est **ce qu'on veut prédire**.

Pour une maison : le **prix**. Pour un email : **spam ou non**. Pour un client : **va-t-il partir ?**

**Analogie :** c'est la réponse à la question qu'on se pose.

### 🏋️ L'"entraînement" (training)

> C'est la phase où le modèle **regarde des milliers d'exemples** pour apprendre.

**Analogie :** comme réviser ses cours avec des annales corrigées. On voit la question ET la réponse, pour comprendre le lien.

### 🧪 Le "test" (évaluation)

> C'est la phase où on vérifie si le modèle a **vraiment compris** — sur des exemples qu'il n'a JAMAIS vus.

**Analogie :** c'est l'examen final. On ne donne pas les réponses cette fois ! Si l'élève a juste mémorisé les annales sans comprendre, il va échouer.

### ⚠️ L'"overfitting" (surapprentissage)

> C'est quand le modèle **apprend par cœur** les exemples d'entraînement, mais est incapable de généraliser.

**Analogie :** un élève qui apprend par cœur les réponses des annales 2020-2023. Il a 20/20 sur ces annales, mais 5/20 à l'examen 2024 car les questions sont différentes. Il a **mémorisé** au lieu de **comprendre**.

### 🎲 Le "train/test split"

> On sépare nos données : par exemple **80% pour entraîner**, **20% pour tester**.

**Analogie :** on révise sur 80% des exemples, et on garde 20% cachés pour l'examen final.

### 🎨 Le "fine-tuning" (affinage, réglage fin)

> C'est quand on prend un modèle **déjà entraîné sur des millions d'exemples généraux**, et qu'on le **réentraîne légèrement** sur un sujet précis, pour le spécialiser.

**Analogie**: Un cuisinier passe son CAP cuisine et travaille plusieurs années dans divers restaurants. Un jour, il est embauché dans un restaurant italien. Il n'a pas à réapprendre la cuisine de A à Z, il doit simplement faire des ajustements.

### 🔍 Pourquoi le fine-tuning existe-t-il ?

Réentraîner un modèle de zéro, c'est **des millions d'euros**, **des milliards de textes** et **des mois de calcul**.

Le fine-tuning permet de **réutiliser** tout ce travail, puis de le spécialiser sur un cas précis avec **quelques milliers d'exemples**, **un simple ordinateur** et **quelques heures ou jours**.

**Analogie :** plutôt que de former un nouveau cuisinier sans expérience à qui il faudra apprendre la cusine de A à Z, vous embaucher un cuisinier qui a déjà 20 ans d'expérience et vous le former **juste** à la cuisine de votre restaurant.

## 💡 À retenir

| Terme | Analogie |
|---|---|
| **Feature** | Fiche descriptive (ex : surface) |
| **Target** | La réponse à prédire (ex : prix) |
| **Entraînement** | Réviser avec annales corrigées |
| **Test** | Passer l'examen sans les réponses |
| **Overfitting** | Apprendre par cœur au lieu de comprendre |

### I) Créer sa base de données

Dans un premier temps nous allons récupérer des images de pizzas sur DuckDuckGo.

```python
urls = search_images('pizza', max_images=1)
urls[0]
```

Nous obtenons un lien ...

```python
from fastdownload import download_url
dest = 'pizza.jpg'
download_url(urls[0], dest, show_progress=False)
```

On le télécharge et on visualise l'image

```python
from fastai.vision.all import *
im = Image.open(dest)
im.to_thumb(256,256)
```

Cherchons et télechargeons des images de brownies

```python
download_url(search_images('brownies', max_images=1)[0], 'brownie.jpg', show_progress=False)
Image.open('brownie.jpg').to_thumb(256,256)
```

Téléchargeons maintenant plusieurs images de pizzas et de brownie

```python
searches = 'brownie','pizza'
path = Path('pizza_ou_brownie')


for o in searches:
    dest = (path/o)
    dest.mkdir(exist_ok=True, parents=True)
    download_images(dest, urls=search_images(f'{o} photo'))
    time.sleep(5)
    resize_images(path/o, max_size=400, dest=path/o)
```

### II) Entrainer le modèle

Nous regroupons maintenant les images dans un Data Block qui contient toutes les informations dont le modèle à besoin...

```python
failed = verify_images(get_image_files(path))
failed.map(Path.unlink)

dls = DataBlock(
    blocks=(ImageBlock, CategoryBlock),
    get_items=get_image_files,
    splitter=RandomSplitter(valid_pct=0.2, seed=42),
    get_y=parent_label,
    item_tfms=[Resize(192, method='squish')]
).dataloaders(path, bs=32)
```

Que contient ce Data Block ?

* *blocks=(ImageBlock, CategoryBlock)* les inputs sont des images, les outputs des catégories. Notez qu'ici les features dont nous avons parlé plus haut sont les caratéristiques des image (formes, couleurs ...)
  
* *get_items=get_image_files* Il s'agit du train/test split dont que nous avions évoqué. On divise les données en 2. 80% des images seront utilisées pour l'entrainement du modèle et 20% pour le tester

* *get_y=parent_label* il s'agit de la cible, la "target" en l'occurence ici il s'agit de pizza ou brownie

Il est possible de visualiser les données dont nous disposons

```python
dls.show_batch(max_n=6)
```

On peut désormais donner toutes ces images avec les labels (bonne réponse) à un modèle. Ici, le modèle est un resnet18.

```python
learn = vision_learner(dls, resnet18, metrics=accuracy)
learn.fine_tune(5)
```

Bravo ! vous avez entrainé votre premier modèle

Mais que veut dire ce tableau ?
* *epoch* correspond correspond à un passage complet de l'ensemble des données à travers l'algorithme
* *train_loss* est le score du modèle lors de son entrainement
* *valid_loss* est le score du modèle lors de sa phase de test
* *accuracy* est l'exactitude, c'est à dire la proportion globale de prédictions correctes

On voit que le modèle atteint une précision de **100%** en quelque époques!

### III) Tester le modèle

On a construit ci dessous une fonction pour chercher des nouvelles images et tester le modèle avec.

**Disclaimer**: Vous n'avez pas à comprendre le code ci dessous, il s'agit seulement d'une fonction qui permet de visuliser des images avant de les tester

```python
import requests
import matplotlib.pyplot as plt
from io import BytesIO
from fastai.vision.all import PILImage


def run_pipeline(query, num_images, learner):
    # 1. Recherche + téléchargement des images valides
    with DDGS() as ddgs:
        results = list(ddgs.images(query, max_results=num_images))
    headers = {"User-Agent": "Mozilla/5.0"}
    collected = []
    for r in results:
        try:
            resp = requests.get(r["image"], headers=headers, timeout=10)
            resp.raise_for_status()
            collected.append({"img": PILImage.create(BytesIO(resp.content)), "url": r["image"]})
        except Exception as e:
            print(f"Image ignorée ({type(e).__name__}) : {r['image'][:60]}...")
    if not collected:
        print("Aucune image valide récupérée.")
        return [], []

    # 2. Petit helper local pour dessiner une grille
    def show_grid(titles, suptitle):
        n = len(collected)
        cols = min(5, n)
        rows = (n + cols - 1) // cols
        fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 4 * rows))
        axes = axes.flatten() if n > 1 else [axes]
        for i, (item, title) in enumerate(zip(collected, titles)):
            axes[i].imshow(item["img"])
            axes[i].set_title(title, fontsize=11)
            axes[i].axis("off")
        for j in range(n, len(axes)):
            axes[j].axis("off")
        plt.suptitle(suptitle, fontsize=14)
        plt.tight_layout()
        plt.show()

    # 3. Aperçu avant prédiction
    print(f"Recherche de {num_images} images pour « {query} »...\n")
    show_grid([f"Image {i+1}" for i in range(len(collected))],
              f"Images trouvées pour « {query} » (avant prédiction)")

    # 4. Prédictions
    print(f"\n{len(collected)} images valides trouvées.")
    predictions = []
    for item in collected:
        pred, idx, probs = learner.predict(item["img"])
        predictions.append({"prediction": pred, "probability": float(probs[idx])})
    show_grid([f"Prédiction : {p['prediction']}\n({p['probability']:.1%})" for p in predictions],
              "Prédictions du modèle")
```

### Le Test
